```python
import pandas as pd
import re
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("data_final.csv")
df.shape
```

```python
df.head(20)
```

```python
a=["city", "building_material", "windows_type", "building_type", "building_floors", "heating_type", "condition", "ownership", "property_type", "rooms"]
for cat in a :
    print(f"{cat}:\n{df[cat].unique()}\n")
```

```python
df['rooms'] = df['rooms'].replace('more', 10)
```

```python
df.drop(['url', 'ad_id', 'district', 'city', 'floor_type', 'rent_cost', 'property_type', 'date_created', 'building_floors', 'floor'], inplace=True, axis=1)
```

```python
print(df['building_material'].value_counts())
```

```python
df['building_material'] = df['building_material'].replace({
    'HYDROTON': 'OTHER',
    'WOOD': 'OTHER'
})

df['building_material'] = df['building_material'].fillna('unknown')
df = df.join(df['building_material'].str.get_dummies(sep=', ').add_prefix('building_material_'))
df.drop(columns=['building_material'], inplace=True)
```

```python
df['ownership'] = df['ownership'].fillna('unknown')
df = df.join(df['ownership'].str.get_dummies(sep=', ').add_prefix('ownership_'))
df.drop(columns=['ownership'], inplace=True)
```

```python
df['condition'] = df['condition'].fillna('unknown')
df = df.join(df['condition'].str.get_dummies(sep=', ').add_prefix('condition_'))
df.drop(columns=['condition'], inplace=True)
```

```python
df['heating_type'] = df['heating_type'].fillna('unknown')
df = df.join(df['heating_type'].str.get_dummies(sep=', ').add_prefix('heating_type_'))
df.drop(columns=['heating_type'], inplace=True)
```

```python
df['building_type'] = df['building_type'].fillna('unknown')
df = df.join(df['building_type'].str.get_dummies(sep=', ').add_prefix('building_type_'))
df.drop(columns=['building_type'], inplace=True)
```

```python
df['windows_type'] = df['windows_type'].fillna('unknown')
df = df.join(df['windows_type'].str.get_dummies(sep=', ').add_prefix('windows_type_'))
df.drop(columns=['windows_type'], inplace=True)
```

```python
# ============================================
# building_year - OPCJA B: Kategorie BEZ flagi
# ============================================

def categorize_age(year):
    if pd.isna(year):
        return 'unknown'
    elif year < 1945:
        return 'prewar'
    elif year < 1990:
        return 'communist'
    elif year < 2010:
        return 'modern'
    else:
        return 'new'

df['building_age_category'] = df['building_year'].apply(categorize_age)

# One-hot encoding
df = pd.get_dummies(df, columns=['building_age_category'], dtype=int)

# Usuń oryginalną kolumnę
df.drop(columns=['building_year'], inplace=True)
```

```python
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
```

```python
train_df.to_csv("train_otodom.csv", index=False)
test_df.to_csv("test_otodom.csv", index=False)
```

```python
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import joblib
import matplotlib.pyplot as plt
# PRZED treningiem sieci, wytrenuj szybki Random Forest żeby sprawdzić importance:
from sklearn.ensemble import RandomForestRegressor

# Przygotuj dane (BEZ skalowania dla RF)
X_temp = train_df.drop(columns=['price'])
y_temp = train_df['price']

# Szybki Random Forest
rf_temp = RandomForestRegressor(n_estimators=50, max_depth=10, random_state=42, n_jobs=-1)
rf_temp.fit(X_temp, y_temp)

# Pokaż TOP 20 cech
feature_importance = pd.DataFrame({
    'feature': X_temp.columns,
    'importance': rf_temp.feature_importances_
}).sort_values('importance', ascending=False)

print("\n=== TOP 20 NAJWAŻNIEJSZYCH CECH ===")
print(feature_importance.head(20))

# Usuń nieistotne cechy (importance < 0.001)
important_features = feature_importance[feature_importance['importance'] > 0.001]['feature'].tolist()
print(f"\nCechy do usunięcia (importance < 0.001): {len(X_temp.columns) - len(important_features)}")

# Zostaw tylko ważne cechy
train_df = train_df[important_features + ['price']]
test_df = test_df[important_features + ['price']]
print(f"Nowa liczba cech: {len(important_features)}")
```
